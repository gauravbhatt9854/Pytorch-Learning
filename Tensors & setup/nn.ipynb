```python
import torch
import torch.nn as nn
import torch.optim as optim

class MyModel(nn.Module):
    def __init__(self):
        super().__init__()
        self.Linear = nn.Linear(1,1)

    def forward(self , x):
        return self.Linear(x)
    

x = torch.tensor([[1.],[2.],[3.],[4.]])
y = torch.tensor([[2.],[4.],[6.],[8.]])


model = MyModel()
loss_fn = nn.MSELoss()

optimiser = optim.Adam(model.parameters() , lr = 0.1)

for epoch in range(20):

# reset the value of grade
    optimiser.zero_grad()

    # y = w*x+b
    # initially w and x value set randomly
    y_pred = model(x)

    loss = loss_fn(y_pred , y)

    # fils the value of grade in x
    loss.backward()

    # check the grade of x and change the value of of w and b (grade tell the direction)
    #( w = w - lr * grade) this is how w - weight changes , also some formula for b- bias

    optimiser.step()

    print(f"epoch {epoch + 1}  Loss : {loss.item():.4f}")

torch.save(model.state_dict() , "linear_model.pth")


```

```python
model = MyModel()
model.load_state_dict(torch.load("linear_model.pth"))
model.eval()
```
