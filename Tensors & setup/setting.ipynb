```python
x = torch.tensor(10. , requires_grad=True)
y = torch.tensor(5.)

z = x * y
z2 = x + y
z_true = torch.tensor(40.)

loss = (z_true - z) ** 2
loss.backward()

print(x.grad)

lr = 0.1

# the suggestion is -50 we always take 10% of suggestion in general
with torch.no_grad():
    x-= (lr * x.grad)

print("final value of x :" , x.item())
print(x.grad)

# without this new grad value of x because of z2 is added in existing so output will be 100 +1= 101 not 1
x.grad.zero_()
z2.backward()
print(x.grad)
```

```python
x = torch.tensor([1.,2.,3.,4.])
y_true = torch.tensor([2.,4.,6.,8.])

w = torch.tensor(0.0 , requires_grad=True)

lr = 0.1
epochs = 20

for epoch in range(epochs):
    # step 1 : find the predict value
    y_predict = x * w

    # step 2 : calculate the loss
    loss = ((y_true - y_predict) ** 2).mean()

    # step 3 : backward and grad value 

    loss.backward()

    with torch.no_grad():
        w-= lr * w.grad

    print(f"Epoch {epoch + 1}: w = {w.item():.4f} , loss = {loss.item():.4f}")
    # if we dont do this it goes in wrong direction 

    w.grad.zero_()

```
